```python
#2.b Read each file into a dataframe
import pandas as pd

aapl = pd.read_csv('data/aapl.csv')
amzn = pd.read_csv('data/amzn.csv')
fb = pd.read_csv('data/fb.csv')
nflx = pd.read_csv('data/nflx.csv')
goog = pd.read_csv('data/goog.csv')
```

```python
#2.c Add a column to each dataframe, called 'ticker', indicating the ticker symbol it is for (Apple's is AAPL). This symbol is how you will look up a stock.  Each file name is the ticker symbol.
aapl['ticker'] = 'AAPL'
amzn['ticker'] = 'AMZN'
fb['ticker'] = 'FB'
nflx['ticker'] = 'NFLX'
goog['ticker'] = 'GOOG'
```

```python
#2.f Using 'fanng.csv', use type conversion to change the 'date' column into a 'datetime' and the 'volume' column into integers.  Then sort by 'date' and 'ticker.'
faang = pd.read_csv('data/faang.csv')
faang['date'] = pd.to_datetime(faang['date'])
faang['volume'] = faang['volume'].astype(int)
```

```python
faang.dtypes
```

```python
faang = faang.sort_values(by=['date', 'ticker'])
faang.head()
```

```python
#2.g Find the seven rows within the highest value for volume.
faang.sort_values(by='volume', ascending=False).head(7)
```

```python
#2.h Use the melt() method to make the 'date' and 'ticker' key identifiers.  Melt the remaining columns so that there are not separated.
faang_melted = faang.melt(id_vars=['date', 'ticker'])
faang_melted.head()
```
