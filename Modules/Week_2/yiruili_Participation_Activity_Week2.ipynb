```python
letters = ['a', 'b', 'c', 'd']

print(letters[int(int('3'*2)//11)])
# int(int('3' * 2 )// 11) equals int(33//11) equals 3, letters[3] evaluates to 'd'.

print(letters[-1])
# -1 refers to the last item, which is 'd'.

print(letters[:2])
# :2 refers to the items before [2], which is ['a', 'b']
```

```python
def function_and(items):
    result = ''
    for i in range(len(items) - 1):
        result = result + items[i] + ', '
    result = result + 'and ' + items[-1]
    return result

spam = ['apples', 'bananas', 'tofu', 'cats']
print(function_and(spam))
```
