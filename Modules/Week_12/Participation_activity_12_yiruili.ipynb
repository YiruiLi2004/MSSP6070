```python
import os
import subprocess
import pandas as pd
import numpy as np
from scipy import stats

repo_url = "https://github.com/YiruiLi2004/MSSP6070.git"
repo_path = "/content/MSSP6070"

if not os.path.exists(repo_path):
    subprocess.run(["git", "clone", repo_url], check=True)

os.chdir(repo_path)

print(os.getcwd())
```

```python
student_data = pd.read_csv(
    "data/Student Performance Data.csv"
)

student_data.head()
```

```python
student_data.shape
```

```python
student_data.columns
```

```python
male_preparation = student_data[
    (student_data["gender"] == "male") &
    (student_data["test preparation course"] == "completed")
]["total score"]

others = student_data[
    ~(
        (student_data["gender"] == "male") &
        (student_data["test preparation course"] == "completed")
    )
]["total score"]
```

```python
print("Male preparation group:")
print("n =", len(male_preparation))
print("Mean =", male_preparation.mean())

print()

print("Others:")
print("n =", len(others))
print("Mean =", others.mean())
```

```python
t_stat, p_value = stats.ttest_ind(
    male_preparation,
    others,
    equal_var=False
)

print("t-statistic:", t_stat)
print("p-value:", p_value)
```

```python
alpha = 0.05

if p_value < alpha:
    print("Reject H0.")
    print("There is a statistically significant difference between males who completed the test preparation course and other students.")
else:
    print("Fail to reject H0.")
    print("There is no statistically significant difference between males who completed the test preparation course and other students.")
```

```python
completed = student_data[
    student_data["test preparation course"] == "completed"
]["total score"]

not_completed = student_data[
    student_data["test preparation course"] == "none"
]["total score"]

print("Completed preparation mean:", completed.mean())
print("No preparation mean:", not_completed.mean())
```

```python
t_stat2, p_value2 = stats.ttest_ind(
    completed,
    not_completed,
    equal_var=False
)

print("t-statistic:", t_stat2)
print("p-value:", p_value2)
```

```python
if p_value2 < 0.05:
    print("There is a statistically significant difference in total scores between students who completed test preparation and those who did not.")
else:
    print("There is no statistically significant difference in total scores between students who completed test preparation and those who did not.")
```
