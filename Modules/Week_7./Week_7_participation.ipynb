```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
sales = pd.read_csv("data/InternetSales.csv", encoding="cp1252")
sales.head()
```

```python
sales[["OrderDate", "SalesAmount"]].head()
```

```python
sales["OrderDate"] = pd.to_datetime(sales["OrderDate"])
```

```python
sales["Year"] = sales["OrderDate"].dt.year
```

```python
yearly_sales = sales.groupby("Year")["SalesAmount"].sum()
yearly_sales
```

```python
yearly_sales.plot(kind="bar")

plt.title("Yearly Internet Sales")
plt.xlabel("Year")
plt.ylabel("Sales Amount")
plt.xticks(rotation=0)

plt.savefig("yearly_sales.pdf", bbox_inches="tight")
plt.show()
```
