
In the context of downloading data from Yahoo Finance using the yfinance library, the period parameter defines the range of data to download and can have the following values:

*    1d, 5d: data for the past 1 or 5 days

*   1mo, 3mo, 6mo: data for the past 1, 3, or 6 months
*   1y, 2y, 5y, 10y, ytd: data for the past 1, 2, 5, or 10 years, or year to date.


*   max: data for the maximum available period.


These values allow you to specify the time frame for which you want to retrieve historical data for a given stock.

#Task 1: Read into seperate data frame the following tickers: Apple, Facebook, Google, and Netflix; use Yahoo finance.

```python
apple_ticker = yf.Ticker("AAPL")
facebook_ticker = yf.Ticker("META")
google_ticker = yf.Ticker("GOOGL")
netflix_ticker = yf.Ticker("NFLX")

Apple = apple_ticker.history(start="2023-01-01", end="2024-01-01").reset_index()
Facebook = facebook_ticker.history(start="2023-01-01", end="2024-01-01").reset_index()
Google = google_ticker.history(start="2023-01-01", end="2024-01-01").reset_index()
Netflix = netflix_ticker.history(start="2023-01-01", end="2024-01-01").reset_index()

print("Apple:", Apple.shape)
print("Facebook/Meta:", Facebook.shape)
print("Google:", Google.shape)
print("Netflix:", Netflix.shape)
```

#Task 2:  Join the individual stock data frames into a single data frame.

```python
Apple["Ticker"] = "AAPL"
Facebook["Ticker"] = "META"
Google["Ticker"] = "GOOGL"
Netflix["Ticker"] = "NFLX"

combined_stocks = pd.concat([Apple, Facebook, Google, Netflix], ignore_index=True)
combined_stocks.head()
```

# Task 3: Create a visualization of the following financial metrics for each stock:  quickRatio, currentRatio, totalRevenue, debtToEquity, revenuePerShare, returnOnAssets, returnOnEquity, grossProfits


```python
metrics = [
    "quickRatio",
    "currentRatio",
    "totalRevenue",
    "debtToEquity",
    "revenuePerShare",
    "returnOnAssets",
    "returnOnEquity",
    "grossProfits"
]

ticker_objects = {
    "Apple": apple_ticker,
    "Facebook/Meta": facebook_ticker,
    "Google": google_ticker,
    "Netflix": netflix_ticker
}

financial_data = {}

for company, ticker in ticker_objects.items():
    info = ticker.info
    financial_data[company] = {
        metric: info.get(metric, np.nan)
        for metric in metrics
    }

financial_metrics = pd.DataFrame(financial_data).T

financial_metrics
```

```python
for metric in metrics:
    plt.figure(figsize=(8, 5))

    sns.barplot(
        x=financial_metrics.index,
        y=financial_metrics[metric]
    )

    plt.title(metric + " by Company")
    plt.xlabel("Company")
    plt.ylabel(metric)
    plt.xticks(rotation=20)

    plt.show()
```

#Task 4: Create a single boxplot for the Closing price for each stock.


```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=combined_stocks,
    x="Ticker",
    y="Close"
)

plt.title("2023 Closing Price by Stock")
plt.xlabel("Stock")
plt.ylabel("Closing Price")

plt.show()
```

# Task 5: Create a daily trading volume visualization for the year 2023 for each stock.


```python
plt.figure(figsize=(12, 6))

sns.lineplot(
    data=combined_stocks,
    x="Date",
    y="Volume",
    hue="Ticker"
)

plt.title("Daily Trading Volume in 2023")
plt.xlabel("Date")
plt.ylabel("Trading Volume")
plt.xticks(rotation=45)

plt.show()
```
