```python
# a. When does the code in a function execute: when the function is defined or called?
# The code inside a function executes when the function is called.
def function_1():
    print("Hello")

function_1()
```

```python
# b. How many global scopes are there in a Python program? How many local scopes?
# A Python program has one global scope, and can have many local scopes. A new local scope is created each time a function is called.
name = "Yirui"

def function_2():
    print(name)

function_2()

print(name)

def function_3():
  master = "mssp"
  print(master)

function_3()
```

```python
# c. What happens to variables in a local scope when the function call returns?
# Variables in a local scope are destroyed when the function call returns.
def function_4():
    x = 1
    print(x)

function_4()
# print(x) will show error.
```

```python
# d. How can you prevent a program from crashing when it gets an error?
# Using try and except statements to handle errors. Also, try/except does not prevent an error from occurring, but tells Python what to do when an error occurs so that the program does not crash.
```

```python
# Complete the practice program (see Sweigart, Chapter Four, The Collatz Sequence Project).
def collatz(number):
    if number % 2 == 0:
        result = number // 2
    else:
        result = 3 * number + 1

    print(result, end=' ')
    return result

number = int(input("Enter number: "))
print(number, end=' ')

while number != 1:
  number = collatz(number)
```
