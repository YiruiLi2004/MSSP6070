```python
import requests
import os
import pandas as pd
import subprocess

repo_url = "https://github.com/YiruiLi2004/MSSP6070.git"
repo_path = "/content/MSSP6070"

if not os.path.exists(repo_path):
    subprocess.run(["git", "clone", repo_url], check=True)

os.chdir(repo_path)
print(os.getcwd())
```

```python
search_term = "ballet"
folder_name = "data/ballet_images"

os.makedirs(folder_name, exist_ok=True)
```

```python
api_url = "https://commons.wikimedia.org/w/api.php"

params = {
    "action": "query",
    "generator": "search",
    "gsrsearch": search_term,
    "gsrnamespace": 6,
    "gsrlimit": 10,
    "prop": "imageinfo",
    "iiprop": "url",
    "iiurlwidth": 800,
    "format": "json"
}

headers = {
    "User-Agent": "MSSP6070-ImageDownloader/1.0"
}

response = requests.get(
    api_url,
    params=params,
    headers=headers
)

response.raise_for_status()

data = response.json()
data
```

```python
pages = data["query"]["pages"]

for page in pages.values():
    print(page["title"])
```

```python
downloaded = []

for page in pages.values():

    title = page["title"]
    image_info = page["imageinfo"][0]

    image_url = image_info.get("thumburl", image_info["url"])

    filename = title.replace("File:", "").replace("/", "_")
    filepath = os.path.join(folder_name, filename)

    image_response = requests.get(
        image_url,
        headers=headers
    )

    image_response.raise_for_status()

    with open(filepath, "wb") as image_file:
        for chunk in image_response.iter_content(100000):
            image_file.write(chunk)

    downloaded.append({
        "title": title,
        "url": image_url,
        "filename": filename
    })

    print("Downloaded:", filename)
```

```python
os.listdir(folder_name)
```

```python
downloaded_df = pd.DataFrame(downloaded)

downloaded_df
```

```python
downloaded_df.to_csv("downloaded_images.csv", index=False)
```
